Yêu cầu:

```python
def CheckDoiXung(s):
 flag=True
 for i in range(len(s)):
    if s[i]!=s[len(s)-i-1]:
        flag=False
        break
 return flag
def main():
 print("Nhập 1 chuỗi:")
 s=input()
 if(CheckDoiXung(s)):
    print("Chuỗi bạn nhập đối xứng")
 else:
    print("Chuỗi bạn nhập không đối xứng")
while True:
 main()
 print("Tiếp không Thím?(c/k):")
 s=input()
 if s=="k":
    break
print("CÁM ƠN THÍM")
```

Câu 4: Các hàm quan trọng trong xử lý chuỗi của Python

Yêu cầu:

Trình bày một số hàm quan trọng trong xử lý Chuỗi của Python

- Nhóm chuyển đổi chữ hoa/ chữ thường

upper(): chuyển toàn bộ chuỗi thành chữ hoa

lower(): chuyển toàn bộ chuỗi thành chữ thường

capitalze(): viết hoa chữ cái đầu trong chuỗi

title(): viết hoa chữ cái đầu tiên của mỗi từ

swapcase(): đảo ngược kiểu chữ (hoa <-> thường)


- Nhóm kiểm tra nội dung chuỗi

isalpha(): kiểm tra chuỗi chỉ chứa chữ cái

isdigit(): kiểm tra chuỗi chỉ chứa chữ số

isalnum(): kiểm tra chuỗi chứa chữ cái và số

ispace(): kiểm tra chuỗi chỉ chứa khoảng trắng

- islower(), isupper(): kiểm tra toàn bộ chuỗi là chữ thường hoặc chữ hoa

find(sub): tìm vị trí đầu tiên của chuỗi con sub

rfind(sub): tìm từ phải sang trái

index(sub): giống find() nhưng báo lỗi nếu không tìm thấy

count(sub): đếm số lần xuất hiện của chuỗi con sub

istrip(), - lstrip(), rstrip(): xóa khoảng trắng bên trái hoặc bên phải

replace(old, new): thay thế chuỗi con old bằng new


- Nhóm tách và nối chuỗi

split(sep): tách chuỗi thành danh sách dựa trên ký tự phân cách sep

join(list): nối các phần tử trong danh sách thành chuỗi


- Nhóm căn chỉnh chuỗi

center(width): căn giữa chuỗi với độ rộng width

ljust(width), rjust(width): căn trái hoặc phải chuỗi








Câu 5: Xử lý chuỗi với các hàm cơ bản

Yêu cầu:

Viết chương trình cho phép nhập vào 1 chuỗi.

 Yêu cầu xuất ra:

- Bao nhiêu chữ IN HOA

- Bao nhiêu chữ in thường

- Bao nhiêu chữ là chữ số

- Bao nhiêu chữ là ký tự đặc biệt

- Bao nhiêu chữ là khoảng trắng

- Bao nhiêu chữ là Nguyên Âm

- Bao nhiêu chữ là Phụ âm


```python
def thong_ke_chuoi(chuoi):
    in_hoa = 0
    in_thuong = 0
    chu_so = 0
    ky_tu_dac_biet = 0
    khoang_trang = 0
    nguyen_am = 0
    phu_am = 0

    nguyen_am_set = set("aeiouAEIOU")

    for c in chuoi:
        if c.isupper():
            in_hoa += 1
        if c.islower():
            in_thuong += 1
        if c.isdigit():
            chu_so += 1
        if c.isspace():
            khoang_trang += 1
        if not c.isalnum() and not c.isspace():
            ky_tu_dac_biet += 1
        if c.isalpha():
            if c in nguyen_am_set:
                nguyen_am += 1
            else:
                phu_am += 1

    print("Số chữ IN HOA:", in_hoa)
    print("Số chữ in thường:", in_thuong)
    print("Số chữ là chữ số:", chu_so)
    print("Số ký tự đặc biệt:", ky_tu_dac_biet)
    print("␣ Số khoảng trắng:", khoang_trang)
    print("Số nguyên âm:", nguyen_am)
    print("Số phụ âm:", phu_am)

# Nhập chuỗi từ người dùng
chuoi_nhap = input("Nhập vào một chuỗi bất kỳ: ")
thong_ke_chuoi(chuoi_nhap)
```

Câu 6: Trích lọc số âm trong chuỗi

Yêu cầu:

Viết một hàm đặt tên là NegativeNumberInStrings(str). Hàm này có đối số truyền vào là một chuỗi bất kỳ, Hãy viết lệnh để xuất ra các số nguyên âm trong chuỗi.

Ví dụ: Nếu nhập vào chuỗi “abc-5xyz-12k9l--p” thì hàm phải xuất ra được 2 số nguyên âm đó là -5 và -12

```python
def NegativeNumberInStrings(s):
    ket_qua = []
    i = 0
    while i < len(s):
        if s[i] == '-' and i + 1 < len(s) and s[i+1].isdigit():
            j = i + 1
            so = ''
            while j < len(s) and s[j].isdigit():
                so += s[j]
                j += 1
            ket_qua.append(int('-' + so))
            i = j  # bỏ qua phần số đã xử lý
        else:
            i += 1
    print("Các số nguyên âm tìm được là:", ket_qua)
```

Câu 7: Tối ưu chuỗi danh từ

Yêu cầu:

Viết chương trình tối ưu Chuỗi danh từ

Một Chuỗi được gọi là tối ưu khi: Không chứa các khoảng trắng dư thừa, các từ cách nhau bởi một khoảng trắng, Ký tự đầu tiên của các từ Viết Hoa

Ví dụ:

Input “ TRần duY thAnH ”

Output “Trần Duy Thanh”


```python
def toi_uu_chuoi_danh_tu(chuoi):
    # Xóa khoảng trắng dư thừa và tách thành các từ
    tu_list = chuoi.strip().split()

    # Viết hoa chữ cái đầu mỗi từ, còn lại viết thường
    tu_chuan = [tu.capitalize() for tu in tu_list]

    # Ghép lại thành chuỗi tối ưu
    ket_qua = ' '.join(tu_chuan)
    return ket_qua

# Nhập chuỗi từ người dùng
chuoi_nhap = input("Nhập chuỗi danh từ: ")
chuoi_toi_uu = toi_uu_chuoi_danh_tu(chuoi_nhap)

print("Chuỗi tối ưu:", chuoi_toi_uu)
```

Câu 8: Tách lấy tên bài hát

Yêu cầu:

Cho một chuỗi là đường dẫn của 1 file nhạc, ví dụ: d:\\music\muabui.mp3

Hãy viết 2 hàm để:

- Lấy ra muabui.mp3

- Lấy ra muabui

Lưu ý đường dẫn bài hát là bất kỳ. Nên khi truyền vào bài hát nào thì lấy chính xác theo bài hát đó.

```python
def lay_ten_file_duoi(path):
    # Tách phần cuối cùng sau dấu "\" hoặc "/"
    ten_file = path.strip().split("\\")[-1]
    return ten_file
def lay_ten_bai_hat(path):
    ten_file = lay_ten_file_duoi(path)
    ten_bai_hat = ten_file.rsplit(".", 1)[0]
    return ten_bai_hat
```
