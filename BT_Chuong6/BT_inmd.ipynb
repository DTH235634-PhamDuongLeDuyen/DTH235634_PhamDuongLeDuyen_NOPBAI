Câu 1: Xử lý List

Yêu cầu:

Viết chương trình cho phép:

- Khởi tạo list

- Thêm phần tử vào list

- Nhập k, kiểm tra k xuất hiện bao nhiêu lần trong list

- Tính tổng các số nguyên tố trong list

- Sắp xếp

- Xóa list

```python
from random import randrange

print("Chương trình xử lý List")
n = int(input("Nhập số phần tử: "))
lst = [0] * n

# Tạo list ngẫu nhiên
for i in range(n):
    lst[i] = randrange(-100, 100)

print("List ngẫu nhiên là:")
print(lst)

# Thêm số mới
value = int(input("Mời bạn thêm số mới: "))
lst.append(value)
print("List sau khi thêm:")
print(lst)

# Đếm số xuất hiện
k = int(input("Bạn muốn đếm số nào: "))
dem = lst.count(k)
print(k, "xuất hiện", dem, "lần trong list")

# Hàm kiểm tra số nguyên tố
def CheckPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

# Đếm và tính tổng số nguyên tố
demnt = 0
tongnt = 0
for x in lst:
    if CheckPrime(x):
        demnt += 1
        tongnt += x

print("Có", demnt, "số nguyên tố trong list")
print("Tổng các số nguyên tố =", tongnt)

# Sắp xếp list
lst.sort()
print("List sau khi sort:")
print(lst)

# Xóa list
del lst
print("List sau khi xóa:")
try:
    print(lst)
except NameError:
    print("List đã bị xóa khỏi bộ nhớ.")
```

Câu 2: Xử lý List nhập ngẫu nhiên

Yêu cầu:

Viết chương trình cho phép:

- Viết lệnh khởi tạo ngẫu nhiên n phần tử cho list

- Gọi k là một số nhập từ bàn phím, hãy xóa tất cả các phần tử có giá trị k tồn tại
trong list

- Kiểm tra list có đối xứng hay không

```python
from random import randrange

lst = []
print("Nhập số phần tử:")
n = int(input())

# Tạo list ngẫu nhiên
for i in range(n):
    lst.append(randrange(0, 100))

print("List sau khi tạo ngẫu nhiên là:")
print(lst)

# Chèn thêm số mới
x = int(input("Mời bạn chèn thêm số mới: "))
lst.append(x)
print("List sau khi chèn:")
print(lst)

# Xóa tất cả số k khỏi list
k = int(input("Mời nhập số để xóa: "))
while k in lst:
    lst.remove(k)

print("List sau khi xóa:")
print(lst)

# Hàm kiểm tra đối xứng
def CheckDoiXung(lst):
    for i in range(len(lst) // 2):
        if lst[i] != lst[len(lst) - i - 1]:
            return False
    return True

# Kiểm tra và in kết quả
if CheckDoiXung(lst):
    print("List đối xứng")
else:
    print("List không đối xứng")
```

Câu 3: Xử lý List Đa chiều

Yêu cầu:

Viết chương trình cho phép:

- Khởi tạo và nhập vào ma trận MxN phần tử ngẫu nhiên

- Xuất dòng bất kỳ nhập từ bàn phím

- Xuất cột bất kỳ từ bàn phím

- Xuất số MAX trong ma trận

```python
from random import randint

# Nhập kích thước ma trận
m = int(input("Nhập số dòng M: "))
n = int(input("Nhập số cột N: "))

# Khởi tạo ma trận MxN với số ngẫu nhiên từ 0 đến 99
matrix = [[randint(0, 99) for _ in range(n)] for _ in range(m)]

print("\nMa trận ngẫu nhiên:")
for row in matrix:
    print(row)

# Xuất dòng bất kỳ
row_index = int(input("\nNhập chỉ số dòng cần xuất (0 đến {}): ".format(m - 1)))
if 0 <= row_index < m:
    print("Dòng thứ", row_index, "là:", matrix[row_index])
else:
    print("Chỉ số dòng không hợp lệ.")

# Xuất cột bất kỳ
col_index = int(input("\nNhập chỉ số cột cần xuất (0 đến {}): ".format(n - 1)))
if 0 <= col_index < n:
    column = [matrix[i][col_index] for i in range(m)]
    print("⬇Cột thứ", col_index, "là:", column)
else:
    print("Chỉ số cột không hợp lệ.")

# Tìm số lớn nhất trong ma trận
max_value = max(max(row) for row in matrix)
print("\n🌟 Giá trị lớn nhất trong ma trận là:", max_value)
```

Câu 7: Viết chương trình nhập vào một dãy các số theo thứ tự tăng, nếu nhập sai quy cách thì yêu cầu nhập lại. In dãy số sau khi đã nhập xong


```python
def nhap_day_tang(n):
    lst = []
    print(f"Nhập {n} số theo thứ tự tăng dần:")
    while True:
        lst = []
        for i in range(n):
            x = int(input(f"Nhập số thứ {i+1}: "))
            lst.append(x)
        # Kiểm tra thứ tự tăng
        if all(lst[i] < lst[i+1] for i in range(n - 1)):
            break
        else:
            print("Dãy không tăng dần. Vui lòng nhập lại.\n")
    return lst

# Nhập số lượng phần tử
n = int(input("Nhập số lượng phần tử: "))
day_so = nhap_day_tang(n)

print("\nDãy số hợp lệ theo thứ tự tăng là:")
print(day_so)
```

Câu 8: Viết chương trình nhập vào một dãy n số thực M[0], M[1],..., M[n-1], sắp xếp dãy số theo thứ tự giảm dần. Xuất ra dãy số sau khi sắp xếp.


```python
print("Nhập số lượng phần tử n:")
n = int(input())

M = []
print("Nhập", n, "số thực:")
for i in range(n):
    x = float(input(f"M[{i}] = "))
    M.append(x)

# Sắp xếp giảm dần
M.sort(reverse=True)

print("\n🔽 Dãy số sau khi sắp xếp giảm dần:")
print(M)
```

Câu 9: Xử lý mảng

Yêu cầu:

Viết chương trình nhập vào một mảng số tự nhiên. Hãy xuất ra màn hình:

- Dòng 1 : gồm các số lẻ, tổng cộng có bao nhiêu số lẻ.

- Dòng 2 : gồm các số chẵn, tổng cộng có bao nhiêu số chẵn.

- Dòng 3 : gồm các số nguyên tố.

- Dòng 4 : gồm các số không phải là số nguyên tố.

M[] ={3,6,7,8,11,17,2,90,2,5,4,5,8}

➔ 3, 7, 11,17, 5(2) ➔6 số lẻ

```python
# Mảng đầu vào
M = [3, 6, 7, 8, 11, 17, 2, 90, 2, 5, 4, 5, 8]

# Hàm kiểm tra số nguyên tố
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

# Phân loại
so_le = [x for x in M if x % 2 != 0]
so_chan = [x for x in M if x % 2 == 0]
so_nguyen_to = [x for x in M if is_prime(x)]
khong_nguyen_to = [x for x in M if not is_prime(x)]

# Xuất kết quả
print("🔹 Dòng 1 - Các số lẻ:", so_le)
print("Tổng cộng có", len(so_le), "số lẻ.\n")

print("🔹 Dòng 2 - Các số chẵn:", so_chan)
print("Tổng cộng có", len(so_chan), "số chẵn.\n")

print("🔹 Dòng 3 - Các số nguyên tố:", so_nguyen_to)
print("🔹 Dòng 4 - Các số không phải nguyên tố:", khong_nguyen_to)
```

Câu 10: Xử lý Ma Trận

Yêu cầu:

Nhập 2 matrix A, B.

Cộng 2 matrix

Viết hàm tính matrix hoán vị➔áp dụng để tìm cho A, B

```python
# Hàm nhập ma trận
def nhap_ma_tran(m, n, ten):
    print(f"\nNhập ma trận {ten} ({m}x{n}):")
    mat = []
    for i in range(m):
        hang = []
        for j in range(n):
            x = int(input(f"{ten}[{i}][{j}] = "))
            hang.append(x)
        mat.append(hang)
    return mat

# Hàm cộng 2 ma trận
def cong_ma_tran(A, B):
    m = len(A)
    n = len(A[0])
    C = []
    for i in range(m):
        hang = []
        for j in range(n):
            hang.append(A[i][j] + B[i][j])
        C.append(hang)
    return C

# Hàm hoán vị ma trận (chuyển vị)
def hoan_vi(ma_tran):
    m = len(ma_tran)
    n = len(ma_tran[0])
    transposed = []
    for j in range(n):
        hang = []
        for i in range(m):
            hang.append(ma_tran[i][j])
        transposed.append(hang)
    return transposed

# Nhập kích thước ma trận
m = int(input("Nhập số dòng: "))
n = int(input("Nhập số cột: "))

# Nhập ma trận A và B
A = nhap_ma_tran(m, n, "A")
B = nhap_ma_tran(m, n, "B")

# Cộng ma trận
C = cong_ma_tran(A, B)
print("\n🔹 Ma trận A + B là:")
for row in C:
    print(row)

# Hoán vị ma trận A
A_T = hoan_vi(A)
print("\n🔄 Ma trận hoán vị của A:")
for row in A_T:
    print(row)

# Hoán vị ma trận B
B_T = hoan_vi(B)
print("\n🔄 Ma trận hoán vị của B:")
for row in B_T:
    print(row)
```
